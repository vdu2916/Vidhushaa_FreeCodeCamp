```python
# Initialize your main data structures
library_books = {}
library_members = {}

def add_book(isbn,title,author,genre,total_copies):
  if not isinstance(isbn,str) or not isbn.strip():
    print("ISBN must be a non-empty string")
    return(False)
  if not isinstance(title,str) or not title.strip():
    print("title must be a non-empty string")
    return(False)
  if not isinstance(author,str) or not author.strip():
    print("author must be a non-empty string")
    return(False)
  if not isinstance(genre,str) or not genre.strip():
    print("genre must be a non-empty string")
    return(False)
  if not isinstance(total_copies,int) or total_copies<=0:
    print("Total copies must be a positive integer")
    return(False)
  if isbn in library_books.keys():
    library_books[isbn]["total_copies"]+=total_copies
    library_books[isbn]["available_copies"]+=total_copies
  else:
    library_books[isbn]={
        "title":title,
        "author":author,
        "genre":genre,
        "total_copies":total_copies,
        "available_copies":total_copies
        }

def add_member(member_ID, name):
  if member_ID in library_members.keys():
    print("Member already exists")
    return(False)
  else:
    library_members[member_ID]={
        "name":name,
        "borrowed_books":set()
    }

def borrow_book(member_id, isbn):
  if member_id not in library_members.keys():
    print("Member does not exist")
    return(False)

  if isbn not in library_books.keys():
    print("Book does not exist")
    return(False)

  if library_books[isbn]["available_copies"]<1:
    print("There are no available copies")
    return(False)
  if isbn in library_members[member_id]["borrowed_books"]:
    print("Member has already borrowed this book")
    return(False)

  library_books[isbn]["available_copies"]-=1
  (library_members[member_id]["borrowed_books"]).add(isbn)
  return(True)

def return_book(member_id, isbn):
  if member_id not in library_members.keys():
    print("Member does not exist")
    return(False)

  if isbn not in library_books.keys():
    print("Book does not exist")
    return(False)

  if isbn not in library_members[member_id]["borrowed_books"]:
    print("Member did not borrow this book")
    return(False)

  library_books[isbn]["available_copies"]+=1
  (library_members[member_id]["borrowed_books"]).remove(isbn)
  return(True)

def display_book_info(isbn):
  if isbn not in library_books.keys():
    return(None)
  else:
    print(f"""ISBN:{isbn}
Title:{library_books[isbn]["title"]}
Author:{library_books[isbn]["author"]}
Genre:{library_books[isbn]["genre"]}
Total Copies:{library_books[isbn]["total_copies"]}
Availble Copies:{library_books[isbn]["available_copies"]}""")

def display_member_info(member_id):
  if member_id not in library_members.keys():
    return(None)
  else:
    print(f"""MemberID:{member_id}
    Name:{library_members[member_id]["name"]}
    Borrowed Books:{library_members[member_id]["borrowed_books"]}""")

def list_all_books():
  for book in library_books:
    print(f"""ISBN:{book}
Title:{library_books[book]["title"]}
Author:{library_books[book]["author"]}
Genre:{library_books[book]["genre"]}
Total Copies:{library_books[book]["total_copies"]}
Availble Copies:{library_books[book]["available_copies"]}""")



print("--- Adding New Books ---")
add_book('978-0321765723', 'The Python Crash Course', 'Eric Matthes', 'Programming', 5)
add_book('978-0134444303', 'Clean Code', 'Robert C. Martin', 'Software Engineering', 3)
add_book('978-1491957669', 'Fluent Python', 'Luciano Ramalho', 'Programming', 2)

# Update existing book (add more copies)
print("\n--- Updating Existing Book ---")
add_book('978-0321765723', 'The Python Crash Course', 'Eric Matthes', 'Programming', 2) # Add 2 more copies

# Test invalid inputs for add_book
print("\n--- Testing Invalid Book Additions ---")
add_book('', 'Invalid Title', 'Author', 'Genre', 1) # Empty ISBN
add_book('978-XXX', '', 'Author', 'Genre', 1)    # Empty Title
add_book('978-YYY', 'Valid Title', ' ', 'Genre', 1) # Whitespace Author
add_book('978-ZZZ', 'Valid Title', 'Author', ' ', 1) # Whitespace Genre
add_book('978-BAD', 'Bad Copies', 'Author', 'Genre', 0) # Zero copies
add_book('978-BAD2', 'Bad Copies 2', 'Author', 'Genre', -1) # Negative copies
add_book('978-BAD3', 'Bad Copies 3', 'Author', 'Genre', 'two') # Invalid type for copies


print("--- Displaying Book Info ---")
display_book_info('978-0321765723')
display_book_info('978-0134444303')
display_book_info('978-9999999999') # Non-existent book

print("\n--- Listing All Books ---")
list_all_books()


print("--- Adding New Members ---")
add_member('M001', 'Alice Smith')
add_member('M002', 'Bob Johnson')
add_member('M003', 'Charlie Brown')

# Try to add an existing member
print("\n--- Adding Existing Member ---")
add_member('M001', 'Alice Smith (Duplicate)')


print("--- Displaying Member Info ---")
display_member_info('M001')
display_member_info('M002')
display_member_info('M004') # Non-existent member


print("--- Testing Borrow Book ---")

# Successful borrows
print("\n--- Successful Borrows ---")
borrow_book('M001', '978-0321765723') # Alice borrows Python Crash Course
borrow_book('M002', '978-0134444303') # Bob borrows Clean Code
borrow_book('M001', '978-1491957669') # Alice borrows Fluent Python

# Verify changes
display_book_info('978-0321765723')
display_member_info('M001')

# Error cases
print("\n--- Error Cases for Borrowing ---")
borrow_book('M999', '978-0321765723') # Non-existent member
borrow_book('M001', '978-9999999999') # Non-existent book
borrow_book('M001', '978-0321765723') # Alice tries to borrow Python Crash Course again (already borrowed)

# Deplete copies and try to borrow
print("\n--- Depleting Copies ---")
borrow_book('M002', '978-0321765723')
borrow_book('M003', '978-0321765723')
borrow_book('M002', '978-1491957669')
borrow_book('M003', '978-1491957669') # Last copy of Fluent Python

print("\n--- Attempting to Borrow Unavailable Book ---")
display_book_info('978-1491957669') # Should show 0 available copies
borrow_book('M001', '978-1491957669') # Alice tries to borrow Fluent Python, should fail

# List all books to see available copies
print("\n--- Library State After Borrows ---")
list_all_books()


print("--- Testing Return Book ---")

# Successful returns
print("\n--- Successful Returns ---")
return_book('M001', '978-0321765723') # Alice returns Python Crash Course
return_book('M002', '978-0134444303') # Bob returns Clean Code

# Verify changes
display_book_info('978-0321765723')
display_member_info('M001')

# Error cases
print("\n--- Error Cases for Returning ---")
return_book('M999', '978-0321765723') # Non-existent member
return_book('M001', '978-9999999999') # Non-existent book
return_book('M001', '978-0134444303') # Alice tries to return Clean Code (not borrowed by her)


print("--- Final Library State ---")
list_all_books()
display_member_info('M001')
display_member_info('M002')
display_member_info('M003')
```
